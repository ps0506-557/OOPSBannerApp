```kotlin
import java.lang.String

object BannerUC3 {
    @JvmStatic
    fun main(args: Array<String>) {
        val O = arrayOf<String?>(
            "  OOOOO  ",
            " O     O ",
            " O     O ",
            " O     O ",
            " O     O ",
            " O     O ",
            "  OOOOO  "
        )

        val P = arrayOf<String?>(
            " PPPPPP  ",
            " P     P ",
            " P     P ",
            " PPPPPP  ",
            " P       ",
            " P       ",
            " P       "
        )

        val S = arrayOf<String?>(
            "  SSSSS  ",
            " S     S ",
            " S       ",
            "  SSSSS  ",
            "       S ",
            " S     S ",
            "  SSSSS  "
        )

        // Print 7 lines
        for (i in 0..6) {
            println(
                String.join(
                    " ",
                    O[i],
                    O[i],
                    P[i],
                    S[i]
                )
            )
        }
    }
}
```
